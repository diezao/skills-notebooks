# Verify matrix output across compute devices

Predict the product of these small synthetic matrices. Run the check before changing the input. If your notebook offers an available CUDA device the cell will use it. The CPU path works without acceleration.

```python
from importlib.util import find_spec

left = [[1, 2], [3, 4], [-1, 5]]
right = [[2, -1], [0, 3]]
prediction = [[2, 5], [6, 9], [-2, 16]]

device = "cpu"
if find_spec("torch") is not None:
    import torch

    device = "cuda" if torch.cuda.is_available() else "cpu"
    a = torch.tensor(left, dtype=torch.float32, device=device)
    b = torch.tensor(right, dtype=torch.float32, device=device)
    observed = (a @ b).cpu().tolist()
else:
    columns = list(zip(*right))
    observed = [[sum(a * b for a, b in zip(row, column)) for column in columns] for row in left]

print("Execution device:", device, "Result:", observed)
assert observed == prediction, "Compare each output with the corresponding row-by-column dot product."
```
