```python
import re

challenge = "sl1:data:0123456789abcdef0123456789abcdef"
assert re.fullmatch(r"sl1:data:[0-9a-f]{32}", challenge), "Enter a data challenge."
seed = challenge.rsplit(":", 1)[1]
```

```python
observations = [int(digit, 16) for digit in seed]
assert len(observations) == 32

# Inspect the frequencies and compute the exact mean as a sum/32 string.
# Keep /32 even when the fraction can be reduced.
from collections import Counter
print(sorted(Counter(observations).items()))
```
