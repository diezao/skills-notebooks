# Predict an instrument count

The readings are synthetic. Choose a threshold and predict how many readings are strictly above it. Run the cell and use its result to refine your prediction. Equality does not count.

```python
from statistics import mean

readings = [18, 22, 21, 28, 19, 30, 24, 26]
threshold = 24
prediction = 3

assert isinstance(threshold, int) and 0 <= threshold <= 100
assert isinstance(prediction, int) and 0 <= prediction <= len(readings)
observed = sum(reading > threshold for reading in readings)
print(f"Mean {mean(readings)}; strict-threshold count {observed}.")
assert prediction == observed, "Recheck the readings that are equal to or above the threshold."
```
