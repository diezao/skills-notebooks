# Evaluate a classifier threshold

Use these synthetic model scores and reference labels to predict the confusion counts at the selected cutoff. Rows at the cutoff are predicted positive. Change the cutoff and decide how a false alarm trades off against a missed case.

```python
scores_and_labels = [(0.88, 1), (0.74, 0), (0.72, 1), (0.60, 1), (0.40, 0), (0.19, 0)]
threshold = 0.70
prediction = {"TP": 2, "FP": 1, "FN": 1, "TN": 2}
assert 0 <= threshold <= 1

counts = dict.fromkeys(("TP", "FP", "FN", "TN"), 0)
for score, label in scores_and_labels:
    predicted = int(score >= threshold)
    category = {(1, 1): "TP", (1, 0): "FP", (0, 1): "FN", (0, 0): "TN"}[predicted, label]
    counts[category] += 1

print(counts)
assert prediction == counts, "Recount predictions at or above the cutoff and compare each with its reference label."
```
