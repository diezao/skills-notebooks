```python
import re

challenge = "sl1:quantum:0123456789abcdef0123456789abcdef"
assert re.fullmatch(r"sl1:quantum:[0-9a-f]{32}", challenge), "Enter a quantum challenge."
seed = challenge.rsplit(":", 1)[1]
```

```python
# First predict whether a Bell pair with the seeded X correction has equal
# bits (even) or different bits (odd). Then compare with the ideal simulator.
from math import sqrt
amplitudes = [1 / sqrt(2), 0, 0, 1 / sqrt(2)]
if int(seed[0], 16) % 2:
    amplitudes = [amplitudes[index ^ 1] for index in range(4)]
for index, amplitude in enumerate(amplitudes):
    print(format(index, "02b"), round(abs(amplitude) ** 2, 8))
```
