Build a confusion matrix from the synthetic labels and model predictions. Return the correct prediction count in the form `correct/32`. Keep the denominator of 32.

```python
import re

challenge = "sl1:ai:0123456789abcdef0123456789abcdef"
assert re.fullmatch(r"sl1:ai:[0-9a-f]{32}", challenge), "Enter an AI evaluation challenge."
seed = challenge.rsplit(":", 1)[1]
```

```python
from collections import Counter

values = [int(digit, 16) for digit in seed]
actual = [int(value % 3 != 0) for value in values]
predicted = [int(value >= 8) for value in values]
counts = Counter(zip(actual, predicted))
for truth in [0, 1]:
    for prediction in [0, 1]:
        print(f"actual={truth} predicted={prediction}: {counts[truth, prediction]}")
```
